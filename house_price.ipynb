```python
import numpy as np

x = np.array([4, 6, 8])        # area in 100s of sq ft  (400, 600, 800 sq ft)
y = np.array([3, 4, 5])        # price in £100k         (£300k, £400k, £500k)

m, c = 0.0, 0.0                # start with a flat line
lr = 0.01                      # learning rate
n = len(x)

for step in range(1, 5001):
    y_pred = m * x + c                    # 1. predict
    error = y_pred - y                    # 2. error
    mse = (error ** 2).mean()             # 3. cost
    dm = (2 / n) * (error * x).sum()      # 4. gradients
    dc = (2 / n) * error.sum()
    m -= lr * dm                          # 5. update (step downhill)
    c -= lr * dc
    if step in (1, 2, 3, 10, 100, 1000, 5000):
        print(f"step {step:>4}: m={m:.3f}  c={c:.3f}  MSE={mse:.4f}")
```
